```python
import tkinter as tk


class clinic_sys:
    def patient_info(self):
        self.name = input("Enter name: ")
        self.surname = input("Enter surname: ")
        self.age = int(input("Enter age: "))
        self.gender = input("Enter gender (F/M): ")

    def survey(self):
        self.Q1 = input("Do you have flu? ")
        self.Q2 = input("Do you have chest pain? ")
        self.Q3 = input("Do you have body discomfort? ")
        self.Q4 = input("Do you take prescriptions? ")


def payment():
    price = {
        "normal": 450,
        "chest": 700,
        "extra": 950
    }

    service = input("Choose service (normal/chest/extra): ")

    if service not in price:
        print("Invalid service selected")
        return None, None, None, None

    selected_price = price[service]

    medical = input("Do you have medical aid? (yes/no): ")

    if medical == "yes":
        eligibility = "Eligible"
        discount = selected_price * 0.15
    else:
        eligibility = "Not Eligible"
        discount = 0

    total = selected_price - discount

    return service, eligibility, discount, total


def show_slip(text):
    window = tk.Tk()
    window.title("Clinic Final Slip")
    window.geometry("400x400")

    label = tk.Label(window, text=text, justify="left", font=("Arial", 10))
    label.pack(padx=10, pady=10)

    window.mainloop()




p1 = clinic_sys()
p1.patient_info()
p1.survey()

service, eligibility, discount, total = payment()

slip = f"""
******** CLINIC FINAL SLIP ********

Name: {p1.name}
Surname: {p1.surname}
Age: {p1.age}
Gender: {p1.gender}

Service Type: {service}
Medical Aid: {eligibility}
Discount: R{discount}
Total: R{total}

******** END ********
"""

file = open("clinic_system.txt", "a")
file.write(slip + "\n")
file.close()

print("Patient details saved successfully!")



show_slip(slip)
```
